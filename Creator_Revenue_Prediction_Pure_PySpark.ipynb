```python
import csv, json
from pathlib import Path
metadata = json.loads(Path('artifacts/metadata.json').read_text(encoding='utf-8'))
with open('artifacts/metrics.csv', encoding='utf-8') as handle:
    metrics = list(csv.DictReader(handle))
print('Engine:', metadata['engine'])
print('Spark:', metadata['spark_version'])
print('Rows:', metadata['rows'], '| Creators:', metadata['unique_creators'])
print('Successful models:', metadata['successful_model_count'])
print('Best model:', metadata['best_model'])
for i, row in enumerate(metrics, 1):
    print(i, row['Model'], '| MAE', f"{float(row['MAE']):,.0f}", '| RMSE', f"{float(row['RMSE']):,.0f}", '| MAPE', f"{float(row['MAPE']):.2f}%", '| R2', f"{float(row['R2']):.4f}")
```
